**Question 2:** In how many different states do the customers live in?

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

states = set()

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # skip header row
    for row in csvreader:
        state = row[4]  # State is the 5th column (index 4)
        states.add(state)

print(f"Number of different states: {len(states)}")
print("All done!")
```

**Question 3** What is the state with most customers?

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

state_counts = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # skip header row
    for row in csvreader:
        state = row[4]
        state_counts[state] = state_counts.get(state, 0) + 1

# find the state with the highest count
top_state = max(state_counts, key=state_counts.get)

print(f"The state with most customers is {top_state} with {state_counts[top_state]} customers")
print("All done!")
```

**Question 4** What is the state with the least customers?

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

state_counts = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # skip header row
    for row in csvreader:
        state = row[4]
        state_counts[state] = state_counts.get(state, 0) + 1

# find the state with the lowest count
bottom_state = min(state_counts, key=state_counts.get)

print(f"The state with the least customers is {bottom_state} with {state_counts[bottom_state]} customers")
print("All done!")
```

**Question 5:** What is the most common last name?

```python
from os.path import exists
import csv
from collections import Counter

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # skip header row
    last_names = [row[2] for row in csvreader]  # Last Name is column index 2

name_counts = Counter(last_names)
top_name, top_count = name_counts.most_common(1)[0]

print(f"The most common last name is {top_name}, appearing {top_count} times")
print("All done!")
```

```python
from os.path import exists
import csv
from collections import Counter, defaultdict
from datetime import datetime

datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    header = next(csvreader)  # CustomerID, OrderID, Date, OrderTotal, ProductName, Price
    rows = list(csvreader)


order_info = {}
for row in rows:
    customer_id, order_id, date, order_total, product, price = row
    if order_id not in order_info:
        order_info[order_id] = {
            'customer': customer_id,
            'date': date.strip(),
            'total': float(order_total)
        }

print("All done loading and cleaning!")
```

```python
print(f"Number of unique orders: {len(order_info)}")
```

```python
items_per_order = Counter(row[1] for row in rows)  # counts rows (product lines) per OrderID
avg_items = sum(items_per_order.values()) / len(items_per_order)
print(f"Average items per order: {round(avg_items, 2)}")
```

```python
max_items = max(items_per_order.values())
print(f"Highest number of items per order: {max_items}")
```

```python
oct_2021_count = 0
for order_id, info in order_info.items():
    if info['date']:  # skip the ~2,142 rows with a missing date
        d = datetime.strptime(info['date'].split('.')[0], '%Y-%m-%d %H:%M:%S')
        if d.year == 2021 and d.month == 10:
            oct_2021_count += 1

print(f"Orders placed in October 2021: {oct_2021_count}")
```

```python
spend_2021 = defaultdict(float)
for order_id, info in order_info.items():
    if info['date']:
        d = datetime.strptime(info['date'].split('.')[0], '%Y-%m-%d %H:%M:%S')
        if d.year == 2021:
            spend_2021[info['customer']] += info['total']  # counted ONCE per order

top_customer = max(spend_2021, key=spend_2021.get)
print(f"Customer {top_customer} spent the most in 2021: ${spend_2021[top_customer]:,.2f}")
```

```python
month_sales = defaultdict(float)
for order_id, info in order_info.items():
    if info['date']:
        d = datetime.strptime(info['date'].split('.')[0], '%Y-%m-%d %H:%M:%S')
        month_sales[d.month] += info['total']

month_names = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
best_month_num = max(month_sales, key=month_sales.get)
print(f"Best month for sales: {month_names[best_month_num - 1]} (${month_sales[best_month_num]:,.2f})")
```
